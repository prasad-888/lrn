```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "gpt2"

# load tokenizer and model

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Input prompt
text = "India is a beautiful country becuase"
inputs = tokenizer(text, return_tensors='pt')

# Generate output
output = model.generate(
    inputs['input_ids'],
    attention_mask=inputs['attention_mask'], # Added to address the warning
    max_length=60,
    temperature=0.8,
    do_sample=True
)

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

## 🎨 Text-to-Image LLMs (Generative Image Models)
-----
Text-to-Image LLMs generate images from natural language prompts. These models convert descriptions like “a futuristic city floating in the clouds” into high-quality images using advanced deep learning techniques.

>💡 Example Output:
Prompt → “A cute golden retriever wearing sunglasses” Image → 🐕😎 (generated by model)

⚙️ How Text-to-Image Models Work\
  🔹 Convert text into embeddings using a text encoder\
  🔹 Condition a diffusion or transformer model on these embeddings\
  🔹 Generate the image by iteratively refining noise → picture\
Many models use Diffusion Models (like Stable Diffusion) which gradually denoise a random latent representation until it becomes a high-quality image.\

🌟 Popular Text-to-Image Models
  🖼️ Stable Diffusion / SDXL — Open-source, customizable\
  🎨 Midjourney — Creative & artistic outputs\
  🧠 DALL·E 3 — Strong language understanding\
  🔬 Google Imagen / Parti — High fidelity (research)\

## Text-to-Speech (TTS)
------
Text-to-Speech (TTS) converts written text into spoken audio. Modern TTS systems produce natural, expressive voices by converting text into acoustic features and then into waveforms using neural vocoders.

⚙️ How TTS Works (Pipeline)
1. Text normalization: expand numbers, abbreviations, dates → normalized text
2. Grapheme → Phoneme (G2P): convert characters to phonemes (pronunciation)
3. Acoustic model: predicts spectrograms or mel-spectrograms from phonemes (e.g., Tacotron2, FastSpeech2)
4. Vocoder: converts spectrograms to waveform audio (e.g., HiFi-GAN, WaveGlow)
5. Post-processing: optional denoising, volume & speed control
🌟 Popular Libraries & Models
  * 🔹 gTTS — simple Google Translate TTS (easy, limited control)
  * 🔹 Coqui TTS — open-source, many voices & training options
  * 🔹 Hugging Face pipelines / models (FastSpeech2, Tacotron2 + HiFi-GAN)
  * 🔹 NVIDIA Tacotron2 + WaveGlow / Triton deploy (high-quality GPU inference)
  * 🔹 Cloud APIs — Amazon Polly, Google Cloud TTS, Azure TTS, OpenAI (commercial, feature-rich)

```python
from transformers import AutoProcessor, VitsModel

import soundfile as sf

from IPython.display import Audio
import torch


processor = AutoProcessor.from_pretrained("facebook/mms-tts-eng")
model = VitsModel.from_pretrained("facebook/mms-tts-eng")


text = "Hello Dhoni! This is a working text to speech audio demo."


inputs = processor(text=text, return_tensors='pt')

with torch.no_grad():
  audio_output = model(**inputs).waveform


audio_data = audio_output.cpu().numpy().squeeze()


sf.write("tts_output.wav", audio_data, model.config.sampling_rate)


Audio("tts_output.wav")
```

> ✅ Tips:
* ✔ End-to-end model — no vocoder needed
* ✔ High-quality, natural speech output
* ✔ Supports many languages (MMS multilingual project)
* ✔ Efficient and fast inference

⚠️ Safety, Licenses & Ethics
  * 🔸 Respect voice copyrights and licensing for pre-trained voices.
  * 🔸 Use consent for synthetic voices of real people.
  * 🔸 Beware of misuse (deepfakes); include watermarks or disclaimers if needed.

🎯 When to Use Which Option
  * gTTS — quick demos, prototypes (internet required)
  * Hugging Face / Coqui — local, customizable, offline options
  * Cloud APIs — scalable, multi-language, production-grade features (SSML, neural styles)

>💡 Pro tip: For best naturalness, pair a modern acoustic model (FastSpeech2/Tacotron2) with a high-quality neural vocoder (HiFi-GAN, WaveGlow). Use SSML or fine-tune for better prosody and speaker control.
