# Introduction to Large Language Models(LLMs)
🤖 What are LLMs?
LLMs (Large Language Models) are a type of artificial intelligence (AI) system trained on massive datasets of text and code. They can perform various natural language tasks such as text generation, translation, summarization, and question answering.

These models consist of millions or even billions of interconnected neurons that recognize complex linguistic patterns. This enables them to learn the statistical relationships between words and phrases and generate new, contextually meaningful text.

📘 Definition and Background
Large Language Models (LLMs) are deep learning models built using transformer-based architectures to process and understand human language. They learn from large text corpora to capture the patterns, semantics, and context of language. Their impact on Natural Language Processing (NLP) has been transformative, enabling machines to exhibit a degree of “language understanding.”

🚀 Applications and Use Cases

📝 Language Generation: Creative writing, storytelling, chatbots.

🌐 Machine Translation: Breaking language barriers across cultures.

❓ Question Answering: Used in virtual assistants and search systems.

📰 Text Summarization: Condenses long text into concise summaries.

💬 Sentiment Analysis: Detects opinions or emotions in text.

🏷️ Named Entity Recognition (NER): Identifies people, organizations, locations, etc.

📚 Language Modeling: Predicts next words in a sentence (core NLP function).

📊 Text Classification: Spam detection, topic categorization, and more.

💡 Key Insight: LLMs are versatile and have revolutionized NLP — powering everything from ChatGPT to machine translation systems.

📜 History of LLMs
The roots of LLMs trace back to early AI research:

🕹️ 1960s – ELIZA: One of the first chatbots that simulated human conversation.

💾 1980s – ARPANET NLU: Early system capable of understanding complex sentences and answering questions.

⚙️ Modern Era: Growth of massive datasets and computing power led to models like GPT, BERT, and T5.


⚖️ Ethical Considerations

# Tokenizer
A Tokenizer is a fundamental component in Natural Language Processing (NLP) that splits raw text into smaller units called tokens. These tokens can be words, subwords, characters, or even symbols, depending on the type of tokenizer and the model being used.

In simpler terms, a Tokenizer converts human language (text) into a form that a machine learning model can understand and process.

When an LLM or NLP model receives text, it cannot process words directly it only understands numerical representations (vectors).

The tokenizer performs the first crucial step in this transformation pipeline


Tokenizer Process:

Raw Text → Tokens → Token IDs → Model Input

Example:

Raw Text: "Machine learning is fun!"

Tokens: ["Machine", "learning", "is", "fun", "!"]

Token IDs: [101, 3698, 4086, 2003, 4569, 999, 102]

Model Input → Passed to Transformer Model

# Functions of a Tokenizer

---



| Function | Description |
|----------|----------|
| Segmentation    | Breaks text into smaller linguistic units (tokens).   |
| Normalization  | Converts text to lowercase, removes punctuation, or handles Unicode normalization.     |
|Mapping|Assigns numerical IDs to tokens using the model’s vocabulary.|
|Padding / Truncation|Adjusts sequences to fixed lengths for batch processing.|
|Detokenization|Converts tokens back into human-readable text.|


## Types of Tokenizers

---


1️⃣ Whitespace Tokenizer\
Splits text by spaces.\
→ Simple but inaccurate for complex languages.

2️⃣ Rule-based Tokenizer\
Uses regular expressions (regex) or grammar rules to detect words and punctuation.

3️⃣ Word Tokenizer\
Treats each word as a token (used in early NLP models).

4️⃣ Subword Tokenizer\
Breaks rare or unknown words into smaller parts (used in modern LLMs).\
Examples:\
🔸 BPE (Byte Pair Encoding) – used in GPT models\
🔸 WordPiece – used in BERT\
🔸 Unigram – used in SentencePiece\

5️⃣ Character Tokenizer\
Treats each character as a token — useful for morphologically rich languages.

## Embedding
--
An Embedding is a mathematical representation of data (like words, sentences, or images) as vectors of real numbers in a continuous, low-dimensional space.

In simple terms:
Embeddings translate meaning into numbers — similar words or ideas get similar vector representations.

image.png

## Why Do We Need Embeddings?
---
Machines cannot understand words or sentences directly — they understand numbers. Embeddings act as a bridge between human language and machine understanding.

💡 Instead of using one-hot encoding (which is sparse and large), embeddings compress information and preserve semantic meaning.\
✅ This allows models to capture relationships like:\
King\
       - Man + Woman ≈ Queen

image.png

⚙️ (Before Embeddings): One-Hot Encoding
--
Before embeddings, words were represented using One-Hot Vectors, where each word in the vocabulary was mapped to a unique binary vector.\

Example Vocabulary:\

["machine", "learning", "is", "fun"]
|Word|	Vector|
|--|--|
|machine|	[1, 0, 0, 0]|
|learning|	[0, 1, 0, 0]|
|is|	[0, 0, 1, 0]|
|fun	|[0, 0, 0, 1]|

🚫 Problems with One-Hot Encoding\
📏 Huge vectors: For a vocabulary of 100,000 words → vector size = 100,000.\
💬 No meaning: “king” and “queen” are as distant as “king” and “banana.”\
⚙️ Sparse representation: Mostly 0s → inefficient for computation.\
🚫 No semantic similarity: Similar words are not close in vector space.\
💡 Insight: One-hot encoding fails to capture meaning or relationships between words — this led to the rise of word embeddings.\

##🌟  The Solution: Embeddings
---
Instead of large, sparse one-hot vectors, embeddings represent each word as a dense vector of real numbers that capture semantic meaning.\

Concept: Words with similar meanings have similar vector representations.\
|Word	|Embedding (dim=3)|
|----|-----|
|king|	[0.7, 0.2, 0.9]|
|queen|	[0.6, 0.3, 0.8]|
|man	|[0.8, 0.1, 0.6]|
|woman	|[0.5, 0.4, 0.7]|

🧠 Key Idea: These numbers are learned by the model — not manually chosen. Words like "king" and "queen" end up close together in vector space, reflecting their similar meanings.

💡 Benefit: Dense embeddings preserve semantic relationships and make computation efficient for large vocabularies.\



## 1️⃣ Word Embeddings — Word2Vec (Static)
---
🔹 Concept:

Each word is represented by a fixed vector — its representation does not change with context. The model learns these vectors based on how often words appear together (co-occurrence) in large text corpora.

```python
!pip install gensim
```

```python
from gensim.models import Word2Vec

sentences = [
  ["i","love","nlp"],
  ["i","love","machine","learning"],
  ["nlp","is", "fun"],
  ["deep","learning","is","powerful"]
]


model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, workers=4)


vector = model.wv["nlp"]
print("Embedding for 'nlp':\n", vector)
```

## 2. Subword Embeddings -FastText
---
🔹 Concept:\
Unlike Word2Vec, FastText represents each word as a combination of subword units (character n-grams). This allows the model to understand internal word structure and generate embeddings even for rare or unseen words.

```python
from gensim.models import FastText

sentences = [
    ["unbelievable", "movie"],
    ["believable","story"],
    ["the", "movie","was","unforgettable"]
    ]


model = FastText(sentences, vector_size=50, window=3, min_count=1)

print("Vector for 'unbelievable':", model.wv['unbelievable'][:5])
print("Vector for unseen word 'unbelievableness':", model.wv['unbelievableness'][:5])
```

##⚙️ How Image Embeddings Work
--
Image embeddings convert visual data (pixels) into a vector of numerical values that capture the image’s features, structure, and content. These vectors enable machines to understand and compare images in a mathematical way.

Input → CNN / Vision Transformer → Flattened Feature Map → Dense Vector (Embedding)\
🔍 Step-by-Step Process\
1️⃣ Input Image: The image (e.g., 224×224×3 pixels) is fed into a neural network.\
2️⃣ Feature Extraction: A model like ResNet or ViT (Vision Transformer) learns to detect edges, shapes, textures, and patterns.\
3️⃣ Flattening: Extracted features are flattened into a fixed-length vector.\
4️⃣ Embedding Vector: The final dense vector (e.g., 512D or 1024D) represents the entire image’s meaning.\

🧠 Example Representation\
|Image	|Embedding (Dim=5)|
|----|-----|
|🐱 Cat|	[0.12, 0.88, 0.47, 0.60, 0.33]|
|🐶 Dog|	[0.15, 0.79, 0.50, 0.63, 0.37]|

💡 Key Idea: Images with similar content (e.g., cats and dogs) will have embeddings that are closer in the vector space.

# What is Self-Attention?

---

Definition:
Self-Attention is a core mechanism in Transformers that allows the model to focus on different parts of the same input sequence when computing the representation of each token.

In simple terms — each word looks at other words in the sentence to understand its contextual meaning.



> 💬 Example:
Sentence: "The cat sat on the mat."
When processing the word "cat", the model also attends to "sat" and "mat" to understand the grammatical and semantic relationship.



>
🔍 Intuition: Self-Attention learns which words are important for each other. For example, in “The animal didn’t cross the street because it was too tired.” — “it” attends strongly to “animal”.


image.png

# ⚙️ How It Works
---
Let’s assume we have an input sentence:

> ["I", "love", "NLP"]
  
Each word is first converted into a vector (embedding). Then, for each word, the model computes three key matrices:

| Matrix|Meaning|
|---|----|
|Q (Query)	|Represents what this word is looking for in other words.|
|K (Key)	|Represents what kind of information this word has.|
|V (Value)	|Contains the actual information carried by the word.|

>Intuition: Each word creates its own Quer\y, compares it with all Keys in the sentence, and uses the resulting attention scores to weight the Values — forming context-aware representations.



```python
import torch
import torch.nn.functional as F

# 🔹 Step 1: Define a simple sentence
sentence = ["I", "love", "Gen AI "]
print("Input sentence:", sentence)

# 🔹 Step 2: Create dummy word embeddings (each word = 4-dimensional vector)
# Normally these come from a real embedding model (like Word2Vec or BERT)
x = torch.tensor([
    [1.0, 0.0, 1.0, 0.0],   # "I"
    [0.0, 2.0, 0.0, 2.0],   # "love"
    [1.0, 1.0, 1.0, 1.0]    # "NLP"
])

print("\nWord Embeddings (input):")
for word, vec in zip(sentence, x):
    print(f"{word}: {vec.numpy()}")

# 🔹 Step 3: Create random weight matrices for Query, Key, and Value
W_Q = torch.randn(4, 4)
W_K = torch.randn(4, 4)
W_V = torch.randn(4, 4)

# Compute Q, K, V
Q = x @ W_Q
K = x @ W_K
V = x @ W_V

# 🔹 Step 4: Compute attention scores (similarity between words)
scores = Q @ K.T  # shape (3x3)

# 🔹 Step 5: Apply scaling and softmax to get normalized attention weights
scaled_scores = F.softmax(scores / (K.shape[-1] ** 0.5), dim=1)

# 🔹 Step 6: Compute final context-aware output
output = scaled_scores @ V

# 🔹 Step 7: Print results clearly
print("\nAttention Weights (how much each word attends to others):")
print(scaled_scores)

print("\nSelf-Attention Output Embeddings (contextualized):")
for word, vec in zip(sentence, output):
    print(f"{word}: {vec.detach().numpy()}")
```

## 🧠 What is Add & Normalize?
----
The Add & Normalize step is a fundamental operation in every Transformer Encoder and Decoder block. It is applied after the Self-Attention layer and again after the Feed-Forward layer.
> 💡 Concept:
Add & Normalize = Residual Connection + Layer Normalization

⚙️ Step-by-Step Explanation
1️⃣ Add: The original input (from the previous layer) is added to the layer’s output — this is called a residual connection.
2️⃣ Normalize: The combined result is then normalized (using LayerNorm) to stabilize training and improve convergence.


> Mathematical Formula:\
Output = LayerNorm(Input + Sublayer(Input))


> ✅ Purpose: Helps retain the original information, stabilize gradients, and prevent vanishing/exploding values. Without this step, deep Transformer networks would be hard to train effectively.

## 🧠  Feed-Forward Layer (FFN)
-----
After the self-attention mechanism provides each token a context-aware embedding, the feed-forward layer (FFN) applies a small neural network to each token independently. This transforms each token’s representation further before passing it to the next layer.

image.png



🔹 Key Details:
• The FFN typically consists of two linear (fully connected) layers with a non-linear activation (e.g., ReLU)
• It is applied position-wise meaning the same network is applied to each token’s embedding separately, without mixing information
• Example dimensions: If token embeddings have size dmodel=512
, the hidden size of the FFN might be dffn=2048
 before projecting back to 512. :contentReference[oaicite:2]{index=2}



 > 💡 Intuition: While self-attention handles “who should we look at”, the FFN handles “what transformation should we apply” to each token’s context-aware vector.

```python
import torch
import torch.nn.functional as F

# Embeddings
V = torch.tensor([
    [1., 0., 1., 0.],   # I
    [0., 2., 0., 2.],   # love
    [1., 1., 1., 1.]    # Gen AI
])

# Self-attention weights (pretend values for visualization)
attention = torch.tensor([
    [0.23, 0.54, 0.23],
    [0.90, 0.06, 0.04],
    [0.55, 0.33, 0.12]
])

# Apply attention (weighted sum)
context = attention @ V

# Feed-Forward layer (simple linear + ReLU)
W1, W2 = torch.randn(4, 8), torch.randn(8, 4)
ffn_output = F.relu(context @ W1) @ W2

print("Context-aware output (after attention):\n", context)
print("\nRefined meaning (after FFN):\n", ffn_output)
```

## 🧠 What is the Output Layer?
-----
The Output Layer is the final stage of a Transformer (or any deep NLP model). It takes the contextual embeddings generated by all previous layers — Self-Attention, Feed Forward, and Add & Norm — and converts them into meaningful predictions such as words, classes, or dense embeddings.

⚙️ In Simple Terms

|Stage |	Purpose|
|------|---------|
|🧩 Input Embeddings|	Represent word meanings as numeric vectors|
|🔁 Self-Attention|	Enables words to understand each other’s context|
|🧱 Feed Forward + Add & Norm	|Refines and stabilizes token-level understanding|
|🏁 Output Layer|	Turns final embeddings into predictions (e.g., next word, class)|


🧩 Transformer Encoder Output Layer
In an Encoder-only model (like BERT), the output layer typically connects to a classification head or embedding head.

> Example: The [CLS] token representation is used for sentence-level classification (e.g., sentiment analysis).

⚙️ Decoder Output Layer
In a Decoder model (like GPT), the output layer uses a Linear layer + Softmax to predict the next token in the sequence.

>💡 Summary: The Output Layer converts internal representations into human-interpretable predictions — whether it’s the next word in a sentence or the sentiment of a review.
