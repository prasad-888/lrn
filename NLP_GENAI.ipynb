```python
# Tensors

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

sentences = [
    "But what about second breakfast?",
    "Don't thik he lknows about second breakfast, Pip.",
    "What about elevensies?"
]

# Tokenization + padding + truncation + convert to tensor

encoded = tokenizer(
    sentences,
    padding=True,
    truncation=True,
    return_tensors ="pt"
)

print("Input Ids", encoded['input_ids'])
print('Attention Mask', encoded['attention_mask'])
print('Tensor shape', encoded['input_ids'].shape)
```

```python
# Seq-Seq
"""Input Sequence → [ Encoder ] → Context Vector → [ Decoder ] → Output Sequence
Example:
"How are you?" → [Encoder] → vector → [Decoder] → "तुम कैसे हो?"
"""

import torch
import torch.nn as nn

src_vocab = {"<sos>":0, "<eos>":1, "how":2, "are":3, "you":4, "?":5 }

tgt_vocab = {"<sos>":0, "<eos>":1,"तुम":2, "कैसे":3, "हो":4, "?":5}

idx2tgt = {v:k for k,v in tgt_vocab.items()}


# 2. Encoder Input

sentences = ["<sos>","how","are","you","?","<eos>"]
src_ids = torch.tensor([src_vocab[w] for w in sentences])



class Encoder(nn.Module):
  def __init__(self):
    super().__init__()
    self.hidden = torch.randn(1, 1, 32)

  def forward(self, x):
    return self.hidden, self.hidden


encoder = Encoder()
h, c = encoder(src_ids)

print("Encoder vector:", h.shape)

translation_map = {
    ("how", "are", "you", "?"): ["तुम", "कैसे", "हो", "?"]
}

decoded = translation_map[("how", "are", "you", "?")]

print("\nTranslated Output:")
print(" ".join(decoded))
```

```python
# Translation (manual implementation)
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load the tokenizer and model
model_name = "Helsinki-NLP/opus-mt-en-hi"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = "How are you?"

# Tokenize the input text
input_ids = tokenizer(text, return_tensors="pt").input_ids

# Generate the translation
generated_ids = model.generate(input_ids)

# Decode the generated IDs back to text
translation = tokenizer.decode(generated_ids[0], skip_special_tokens=True)

print(translation)
```
